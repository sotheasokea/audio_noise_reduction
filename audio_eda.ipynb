```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
```

```python
fs, x = wavfile.read("hci_audio_record.wav")
t = np.arange(len(x)) / fs

print("fs =",fs)
print("Samples =",len(x))
print("Duration =",len(x)/fs,"s")
```

```python
plt.figure(figsize = (12, 5))
plt.plot(t, x)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
```

```python
duration = 30
x_30_s = x[:fs*duration]
t_30_s = t[:fs*duration]

plt.figure(figsize = (12, 5))
plt.plot(t_30_s, x_30_s)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
```

```python
duration = 1
x_1_s = x[:fs*duration]
t_1_s = t[:fs*duration]

plt.figure(figsize = (12, 5))
plt.plot(t_1_s, x_1_s)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
```

```python
duration = 0.5
x_5_s = x[:int(fs*duration)]
t_5_s = t[:int(fs*duration)]

plt.figure(figsize = (12, 5))
plt.plot(t_5_s, x_5_s)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
```
